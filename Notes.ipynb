### the objective of this project is to get expertise in SQL optimisation through practical hands on using relatively bigger data sets

### Datasets being used in the beginning (SF - Scale factor)
1. TPC-DS SF1 - Primary Learning Dataset (size 1 GB)
1. TPC-DS SF1000 - Primary Learning Dataset (size 1000 GB / 1 TB)


**TPC-DS** stands for Transaction Processing Performance Council - Decision Support. It is an industry-standard benchmark (a standardized test used to measure system performance) that simulates the complex data of a large retail and e-commerce business.

* **`tpcds_sf1`**: The "sf" stands for Scale Factor. A scale factor of 1 represents approximately **1 Gigabyte (GB)** of data. It is perfect for writing queries and initial testing on a Databricks Free Community Edition workspace.
* **`tpcds_sf1000`**: This represents a scale factor of 1,000, which equals roughly **1 Terabyte (TB)** of data. This massive volume is used for comprehensive performance evaluations and benchmarking on larger Professional or Enterprise clusters.

**The Analogy:**
Think of TPC-DS as a mock retail company's record room. `tpcds_sf1` is the accounting ledger for a single small branch—easy to scan quickly. `tpcds_sf1000` is the global corporate archive containing decades of worldwide records; answering a question here requires significant processing power.

**A Concrete Example:**
Both datasets share identical schemas (such as tables for `store_sales`, `customer`, and `item`). If you run an analytical query (a complex data analysis question) to calculate total regional sales on `sf1`, it finishes quickly. The exact same query on `sf1000` forces Databricks to distribute the workload across multiple computers to process billions of rows.
